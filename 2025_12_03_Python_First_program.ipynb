```python
# First, ensure numeric columns by converting or selecting only numeric columns
# Option 1: Convert all possible columns to numeric
df = df.apply(pd.to_numeric, errors='coerce')  # 'coerce' will convert non-convertible values to NaN

# Option 2: Or select only numeric columns
# numeric_df = df.select_dtypes(include=['number'])

# Now calculate IQR on numeric data
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
df = df[~((df < (Q1 - 1.5 * IQR)) |(df > (Q3 + 1.5 * IQR))).any(axis=1)]
df.shape
```

```python
# First, check if the DataFrame exists and has data
if 'df' in globals() and not df.empty:
    # Check if 'Make' column exists
    if 'Make' in df.columns:
        # Check if the column has non-null values
        if df['Make'].notna().any():
            df.Make.value_counts().nlargest(40).plot(kind='bar', figsize=(10,5))
            plt.title("Number of cars by make")
            plt.ylabel('Number of cars')
            plt.xlabel('Make')
        else:
            print("The 'Make' column contains only null values.")
    else:
        print("The DataFrame does not have a 'Make' column. Available columns are:", df.columns.tolist())
else:
    print("The DataFrame 'df' is not defined or is empty.")
    
# Make sure matplotlib is imported
import matplotlib.pyplot as plt

df.Make.value_counts().nlargest(40).plot(kind='bar', figsize=(10,5))
plt.title("Number of cars by make")
plt.ylabel('Number of cars')
plt.xlabel('Make');
```
