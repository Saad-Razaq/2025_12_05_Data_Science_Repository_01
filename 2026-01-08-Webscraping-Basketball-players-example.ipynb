```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Use requests with a proper User-Agent header to avoid being blocked
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

URL = 'https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)'

# Make the request with headers
response = requests.get(URL, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Remove hyperlinks from the HTML before parsing with pandas
for a in soup.find_all('a'):
    a.replace_with(a.text)

# Use pandas to read the modified HTML content
tables = pd.read_html(str(soup))

# Find the correct table (you might need to adjust the index)
df = tables[2]  # the required table will have index 2

# Clean up any remaining hyperlink artifacts if needed
# This applies a string cleaning function to all string columns
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str).str.replace(r'\[.*?\]', '', regex=True)

print(df)
```
